```python
import pandas as pd
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

print("First 5 rows of dataset:")
print(df.head())

# Mean
mean_values = df.mean()
print("\nMean values:")
print(mean_values)

# Median
median_values = df.median()
print("\nMedian values:")
print(median_values)

# Mode
mode_values = df.mode().iloc[0]
print("\nMode values:")
print(mode_values)
```

```python
import pandas as pd
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

print("First 5 rows of dataset:")
print(df.head())

# Range
range_values = df.max() - df.min()
print("\nRange values:")
print(range_values)

# Variance
variance_values = df.var()
print("\nVariance values:")
print(variance_values)

# Standard Deviation
std_values = df.std()
print("\nStandard Deviation values:")
print(std_values)

# Interquartile Range
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1

print("\nInterquartile Range (IQR):")
print(IQR)
```

```python
import pandas as pd
from sklearn.datasets import load_iris
from scipy.stats import spearmanr, chi2_contingency

# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Pearson correlation
print("Pearson Correlation:")
print(df.corr(method='pearson'))

# Spearman rank correlation
print("\nSpearman Correlation:")
print(df.corr(method='spearman'))

# Chi-square test between categorical variables
table = pd.crosstab(iris.target, df['sepal width (cm)'] > df['sepal width (cm)'].median())

chi2, p, dof, expected = chi2_contingency(table)

print("\nChi-Square Test:")
print("Chi-square statistic:", chi2)
print("P-value:", p)
print("Degrees of freedom:", dof)
```

```python
from sklearn.datasets import load_iris
from scipy.stats import ttest_ind, f_oneway

# Load Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# T-test: compare sepal length of two species
setosa = X[y == 0, 0]
versicolor = X[y == 1, 0]

t_stat, t_p = ttest_ind(setosa, versicolor)

print("T-test:")
print("T-statistic:", t_stat)
print("P-value:", t_p)

# ANOVA: compare sepal length across all three species
setosa = X[y == 0, 0]
versicolor = X[y == 1, 0]
virginica = X[y == 2, 0]

f_stat, anova_p = f_oneway(setosa, versicolor, virginica)

print("\nANOVA:")
print("F-statistic:", f_stat)
print("P-value:", anova_p)
```

```python
import pandas as pd
from scipy.stats import zscore
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Calculate Z-scores
z_scores = df.apply(zscore)

# Identify values beyond +/- 3 standard deviations
outliers = (z_scores.abs() > 3)

print("Outlier counts in each column:")
print(outliers.sum())

print("\nRows containing outliers:")
print(df[outliers.any(axis=1)])
```

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = (df < lower_bound) | (df > upper_bound)

print("Outlier counts in each column:")
print(outliers.sum())

print("\nRows containing outliers:")
print(df[outliers.any(axis=1)])
```
